```python
import bloodline as bl

import pandas as pd
```

```python
left = pd.DataFrame({"id": [1, 2], "a": [10, 20]})
right = pd.DataFrame({"id": [2], "b": [99]})

data_source = bl.Source(
    source_type=bl.get_source_type(name="DATA_SOURCE"), source_metadata={"path": "data/iris.csv", "format": "csv"}
)
left = bl.update_table_data_lineage(left, default_source=data_source)
right = bl.update_table_data_lineage(right)

# Use the pandas accessor
out = left.lineage.merge(right, on="id", how="left")

out["c"] = out["a"] * 2
out.lineage.impute(default_source=bl.Source.rule(name="double_value"))

from rich import print

print(out["data_lineage"].to_dict())
```
