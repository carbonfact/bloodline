```python
import pandas as pd
import bloodline as bl
from rich import print

lineage = bl.Lineage(
    default_source=bl.Source.unknown(reason="default"),
    extra_sources_type=("HEURISTIC", "RULE"),
)

@lineage(metadata={"owner": "data_team"})
def load_products():
    return pd.read_csv("sample_products.csv")

@lineage
def load_users():
    return pd.read_csv("sample_users.csv")

@lineage
def load_purchases():
    purchases = pd.read_csv("sample_purchases.csv")
    users = load_users()
    merged = pd.merge(left=purchases, right=users, left_on="user_id", right_on="id")
    products = load_products()
    merged = pd.merge(left=merged, right=products, left_on="item_sku", right_on="sku")
    return merged

purchases = load_purchases()
```
